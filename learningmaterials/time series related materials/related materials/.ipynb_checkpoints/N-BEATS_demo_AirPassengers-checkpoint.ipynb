```python
import matplotlib.pyplot as plt
from darts.datasets import AirPassengersDataset
from darts.models import NBEATSModel
from darts import TimeSeries
from darts.metrics import mape, mase
```

```python
series: TimeSeries = AirPassengersDataset().load()
series.plot(label="Passenger Count")
plt.title("Monthly Air Passengers")
plt.show()
```

```python
train, val = series.split_before(0.8)
```

```python
model = NBEATSModel(
    input_chunk_length=24,      # look-back window of 24 months
    output_chunk_length=12,     # forecast horizon of 12 months
    generic_architecture=True,  # use generic N-BEATS blocks
    num_stacks=30,
    num_blocks=1,
    num_layers=4,
    layer_widths=256,
    random_state=0,
)
```

```python
model.fit(
    train,
    epochs=100,
    verbose=True,
)
```

```python
forecast = model.predict(n=36)
```

```python
plt.figure(figsize=(10, 6))
train.plot(label="Train")
series.plot(label="Actual")
forecast.plot(label="Forecast")
plt.title("N-BEATS Forecast on AirPassengers")
plt.legend()
plt.show()
```
