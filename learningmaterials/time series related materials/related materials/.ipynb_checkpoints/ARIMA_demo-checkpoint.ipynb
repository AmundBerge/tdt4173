This Jupyter notebook demonstrates how to fit an ARIMA model on a time series dataset using Python's `statsmodels` library. It covers:

1. Loading and visualizing a dataset
2. Checking stationarity
3. Identifying ARIMA orders (p, d, q)
4. Fitting the ARIMA model
5. Diagnostic plots
6. Forecasting future values

# 1. Import Libraries and Load Data

We will use a built-in dataset (`AirPassengers`) for demonstration. You can replace this with your own time series.

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.arima.model import ARIMA
import warnings
warnings.filterwarnings("ignore")
```

```python
url = 'https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv'
data = pd.read_csv(url, parse_dates=['Month'], index_col='Month')
series = data['Passengers']
```

```python
plt.figure(figsize=(10, 4))
series.plot(title='Monthly Air Passengers')
plt.ylabel('Passengers')
plt.show()
```

# 2. Stationarity Check

Perform an Augmented Dickey-Fuller test to check for stationarity.

```python
result = adfuller(series)
print('ADF Statistic: %f' % result[0])
print('p-value: %f' % result[1])
for key, value in result[4].items():
    print(f'Critical Value ({key}): {value:.3f}')
```

The p-value indicates whether the series is stationary. If not, we difference the data.

# 3. Differencing to Achieve Stationarity

First-order differencing removes the trend.

```python
diff_series = series.diff().dropna()
plt.figure(figsize=(10, 4))
diff_series.plot(title='Differenced Series (1st Order)')
plt.show()
```

ADF test on differenced data

```python
result_diff = adfuller(diff_series)
print('Differenced ADF Statistic: %f' % result_diff[0])
print('Differenced p-value: %f' % result_diff[1])
```

# 4. ACF and PACF Plots

Use ACF/PACF plots to identify q and p.

```python
fig, axes = plt.subplots(1, 2, figsize=(15,4))
plot_acf(diff_series, ax=axes[0], lags=30)
plot_pacf(diff_series, ax=axes[1], lags=30)
plt.show()
```

The ACF plot shows autocorrelation of the series at different lags:
- A slow decay in ACF suggests a non-stationary series (before differencing).
- For the differenced (stationary) series, look for:
  - ACF cutoff after lag q implies an MA(q) component.
  - Exponential decay or sinusoidal pattern suggests AR terms.

The PACF plot shows partial autocorrelation:
- A sharp cutoff after lag p indicates an AR(p) component.
- A gradual decay suggests MA terms.

Interpretation example (for the differenced series):
- If PACF cuts off after lag 2, p ≈ 2.
- If ACF cuts off after lag 2, q ≈ 2.

# 5. Fitting the ARIMA Model

Based on ACF & PACF, we choose p=2, d=1, q=2 (for example).

```python
model = ARIMA(series, order=(2, 1, 2))
fitted = model.fit()
print(fitted.summary())
```

# 6. Diagnostic Plots

Diagnostic plots help assess model fit and validate assumptions:

1. **Standardized Residuals**: Should look like white noise (no obvious patterns).
2. **Histogram + KDE**: Residuals should be approximately normally distributed.
3. **QQ Plot**: Points should lie on the 45° line if residuals are normal.
4. **Correlogram (ACF of Residuals)**: No significant autocorrelations should remain.

If residuals show patterns, consider revising p, d, q orders or adding seasonal components.

```python
fitted.plot_diagnostics(figsize=(10, 8))
plt.show()
```

# 7. Forecasting

Forecast the next 12 months and plot the results.

```python
forecast = fitted.get_forecast(steps=12)
mean_forecast = forecast.predicted_mean
conf_int = forecast.conf_int()

plt.figure(figsize=(10, 4))
series.plot(label='Historical')
mean_forecast.plot(label='Forecast')
plt.fill_between(conf_int.index,
                 conf_int.iloc[:, 0],
                 conf_int.iloc[:, 1], alpha=0.3)
plt.legend()
plt.title('Air Passengers Forecast')
plt.show()
```
